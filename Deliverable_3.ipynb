## Task 1 – Update the Character class
`attack()` now calls `target.take_damage()` instead of changing `target._hp` directly.
New: `take_damage()`, `heal()` and the read-only property `is_alive`. HP is always kept between 0 and `hp_max`.

```python
alice = Character("Alice")
bob = Character("Bob", hp_max=50, damage=15)
alice.introduce()
print(alice)
print(repr(bob))
```

```python
bob.attack(alice)
print("After Bob attacks:", alice.hp)
alice.heal(100)          # cannot heal above hp_max
print("After healing 100:", alice.hp)
alice.take_damage(50)   # cannot go below 0
print("After 50 damage:", alice.hp, "| alive:", alice.is_alive)
```

```python
# is_alive is read-only
try:
    alice.is_alive = True
except AttributeError as e:
    print("AttributeError:", e)
```

**Discussion (Task 1):** Bob deals 15 damage, so Alice goes from 100 to 85 HP. Healing 100 only brings her back to 100,
and 500 damage stops at 0, so HP always stays between 0 and `hp_max`. Since `attack()` now uses `take_damage()`,
each object is the only one changing its own `_hp`, which is better encapsulation. Trying to set `is_alive`
gives an `AttributeError`, which shows that the property is read-only.

## Task 2 – Dice class
An n-sided dice (n ≥ 2, default 6) with `roll()`, `__repr__` and `__call__`, so `d20()` gives the same result as `d20.roll()`.

```python
d6 = Dice()
d20 = Dice(sides=20)
print(d6, d20)
print("d20.roll():", d20.roll())
print("d20():     ", d20())
print("10 rolls of d6:", [d6() for _ in range(10)])

try:
    Dice(sides=1)
except ValueError as e:
    print("ValueError:", e)
```

```python
# The average of many d6 rolls should be close to 3.5
rolls = [d6() for _ in range(10_000)]
print("Mean of 10,000 d6 rolls:", sum(rolls) / len(rolls))
print("Min/max:", min(rolls), max(rolls))
```

**Discussion (Task 2):** `repr` gives `Dice(sides=6)`, as required. Over 10 000 rolls of a d6 the mean is about 3.5,
which is the expected value (1+2+…+6)/6, and all rolls are between 1 and 6. A dice with fewer than 2 sides
raises a `ValueError`.

## Task 4 - Character specialisation

```python
class Warrior(Character):
    """Defensive specialist: armor reduces incoming damage."""

    def __init__(self, name, hp_max=100, damage=10, weapon=None, armor=5):
        super().__init__(name, hp_max, damage, weapon)
        self.armor = armor

    def take_damage(self, amount):
        super().take_damage(max(amount - self.armor, 0))


class Mage(Character):
    """Resource-based attacker: spells cost 10 mana and deal double damage."""

    def __init__(self, name, hp_max=100, damage=10, weapon=None, mana=30, mana_max=30):
        super().__init__(name, hp_max, damage, weapon)
        self.mana_max = mana_max
        self._mana = min(mana, mana_max)

    @property
    def mana(self):
        return self._mana

    def attack(self, target):
        if self._mana >= 10:
            self._mana -= 10
            dmg = 2 * self.damage       # spell: ignores weapon
            target.take_damage(dmg)
            return dmg
        dealt = super().attack(target)  # base attack, uses weapon if equipped
        self._mana = min(self._mana + 5, self.mana_max)
        return dealt
```

## Task 5 - Running battles

```python
from character import Character


class Battle:
    """Manages turn-based combat between a hero and an enemy."""

    def __init__(self, hero: Character, enemy: Character) -> None:
        self.hero = hero
        self.enemy = enemy
        self.rounds = 0

    def play_round(self) -> None:
        """Hero attacks enemy; if enemy is still alive, enemy attacks hero."""
        self.rounds += 1
        dealt = self.hero.attack(self.enemy)
        print(f"Round {self.rounds}: {self.hero.name} hits {self.enemy.name} "
              f"for {dealt} (HP {self.enemy.hp}/{self.enemy.hp_max})")
        if self.enemy.is_alive:
            dealt = self.enemy.attack(self.hero)
            print(f"         {self.enemy.name} hits {self.hero.name} "
                  f"for {dealt} (HP {self.hero.hp}/{self.hero.hp_max})")

    def simulate_battle(self) -> Character:
        """Run rounds until one character is dead. Returns the winner."""
        while self.hero.is_alive and self.enemy.is_alive:
            self.play_round()
        winner = self.hero if self.hero.is_alive else self.enemy
        print(f"{winner.name} wins after {self.rounds} rounds!")
        return winner
```
